```python
class FeatureExtractionBlock(nn.Module):
    def __init__(self, in_channels, out_channels, num_conv):
        self.featurebxtractionblock = nn.Sequential()

        for n in range(num_conv):
            self.featurebxtractionblock.append(nn.Conv2d(in_channels if n==0 else out_channels, out_channels, kernel_size=3, padding=1, stride=1))
            self.featurebxtractionblock.append(nn.BatchNorm2d(out_channels))
            self.featurebxtractionblock.append(nn.ReLU(in_place=True))
        self.featurebxtractionblock.append(nn.MaxPool2d(2, 2))

    def forward(self, x):
        return self.featurebxtractionblock(x)
```
