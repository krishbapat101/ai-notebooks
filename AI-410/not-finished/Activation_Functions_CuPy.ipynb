```python
import cupy as xp
import torch
```

```python
class Sigmoid:
  def __init__(self):
    pass

  def forward(self, x):
    return 1/(1 + xp.exp(-x))

  def __call__(self, x):
    return self.forward(x)
```

```python
class ReLU:
  def __init__(self):
    pass

  def forward(self, x):
    return xp.maximum(0, x)

  def __call__(self, x):
    return self.forward(x)
```

```python
class Identity:
  def __init__(self):
    pass

  def forward(self, x):
    return x

  def __call__(self, x):
    return self.forward(x)
```

```python
class Tanh:
    def __init__(self):
        pass

    def forward(self, x):
        return (xp.exp(x) - xp.exp(-x)) / (xp.exp(x) + xp.exp(-x))
```

```python
class Embedding:
    def __init__(self, ne, ed):
        self.w = xp.random.randn(ne, ed)

    def forward(self, idx):
        self.idx = idx
        self.out = self.w[idx]
        return self.out

    def backward(self, grad):
        self.dw = xp.zeros_like(self.w)
        xp.add.at(self.dw, self.idx, grad)
    def parameters(self):
        return [self.w]
```

```python
xp.random.seed(42)
x = xp.random.randn(4, 8)
xt = torch.tensor(x.get(), device="cuda")

print("Identity:", Identity().forward(x).get() - torch.nn.Identity()(xt).cpu().numpy())
print("ReLU:", ReLU().forward(x).get() - torch.relu(xt).cpu().numpy())
print("Sigmoid:", Sigmoid().forward(x).get() - torch.sigmoid(xt).cpu().numpy())
print("Tanh:", Tanh().forward(x).get() - torch.tanh(xt).cpu().numpy())

emb = Embedding(27, 10)
idx = xp.random.randint(0, 27, (4, 3))
wt = torch.tensor(emb.w.get(), device="cuda")
print("Embedding:", emb.forward(idx).get() - wt[torch.tensor(idx.get(), device="cuda")].cpu().numpy())
```
