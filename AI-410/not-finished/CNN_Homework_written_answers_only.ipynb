Output size formula:

$$O = \frac{N + 2P - K}{S} + 1$$

Setting $O = N$ requires:

$$S = 1, \qquad P = \frac{K - 1}{2}$$


=


$$S = 2, \qquad P = \frac{K - 1}{2}, \qquad K \text{ odd}$$

example, $K = 3,\ S = 2,\ P = 1$ gives $O = \lceil N/2 \rceil$.

or

Same-size conv ($S = 1,\ P = \frac{K-1}{2}$) followed by $2 \times 2$ max pooling with stride $2$:

$$O = \frac{N}{2}$$
