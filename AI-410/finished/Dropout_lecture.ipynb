```python
import torch
import torch.nn as nn
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
class Mydropout(nn.Module):
    def __init__(self, p=0.5):
        super().__init__()
        self.p = p
        self.training = True
    def forward(self, X):
        if self.training:
            no_drop = (torch.rand_like(X) > self.p).to(dtype=torch.float32) # mask
            return X * no_drop/ (1 - self.p)
        else:
            return X
```

```python
num_samples = 10
num_features = 5
X = torch.randn(num_samples, num_features)

prob_dropout = 0.3
```

```python
my_model = Mydropout(p=prob_dropout)
my_model.to(device)
X.to(device)
my_model.train()
print(my_model.training)

X_dropped = my_model(X)
print(X_dropped/ X)


# Evaluation mode
my_model.eval()

print(my_model.training)
X_dropped = my_model(X)
print(X_dropped/X)
```

```python
model = nn.Dropout(p = prob_dropout)
```

```python
model.to(device)
model.train()

X_dropped = model(X)
print(X_dropped/ X)


# Evaluation mode
model.eval()

print(my_model.training)
X_dropped = model(X)
print(X_dropped/X)
```
