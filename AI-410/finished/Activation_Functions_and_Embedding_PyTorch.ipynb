```python

import torch
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}\n")
```

```python
class Sigmoid:
  def __init__(self):
    pass

  def forward(self, x):
    return 1/(1 + torch.exp(-x))

  def __call__(self, x):
    return self.forward(x)
```

```python
class linear:
  def __init__(self, weight, ):
    pass

  def forward(self, x):
    return

  def __call__(self, x):
    return self.forward(x)
```

```python
class ReLU:
  def __init__(self):
    pass

  def forward(self, x):
    return torch.clamp(x, min=0)

  def __call__(self, x):
    return self.forward(x)
```

```python
class Identity:
  def __init__(self):
    pass

  def forward(self, x):
    return x

  def __call__(self, x):
    return self.forward(x)
```

```python
class Tanh:
    def __init__(self):
        pass

    def forward(self, x):
        return (torch.exp(x) - torch.exp(-x)) / (torch.exp(x) + torch.exp(-x))

    def __call__(self, x):
        return self.forward(x)
```

```python
class Embedding:
    def __init__(self, ne, ed):
        self.w = torch.randn(ne, ed)

    def forward(self, idx):
        self.idx = idx
        self.out = self.w[idx]
        return self.out

    def backward(self, grad):
        self.dw = torch.zeros_like(self.w)
        self.dw.index_put_((self.idx,), grad, accumulate=True)

    def parameters(self):
        return [self.w]

    def __call__(self, idx):
        return self.forward(idx)
```

```python
x = torch.randn(64, 32, device=device)
idx = torch.randint(0, 100, (64,), device=device)

print("Sigmoid:", Sigmoid()(x) - torch.sigmoid(x))
print("ReLU:   ", ReLU()(x) - torch.relu(x))
print("Tanh:   ", Tanh()(x) - torch.tanh(x))

emb = Embedding(100, 16)
emb.w = emb.w.to(device)
ref = torch.nn.Embedding(100, 16).to(device)
ref.weight.data = emb.w.clone()
print("Emb fwd:", emb(idx) - ref(idx))

grad = torch.randn(64, 16, device=device)
emb.backward(grad)
ref(idx).backward(grad)
print("Emb bwd:", emb.dw - ref.weight.grad)
```
