```python
import torch
import torch.nn as nn
num_samples = 10
input_dim = 5
output_dim = 3
X = torch.randn(size = (num_samples, input_dim))
linear = nn.Linear(in_features=input_dim, out_features=output_dim)
print(linear.weight.shape)
print(linear.bias.shape)

Y_linear = linear(X)

conv1d = nn.Conv1d(in_channels = input_dim, out_channels=output_dim, kernel_size=1, stride =1, padding=0, bias=True)
print(conv1d.weight.shape)
print(conv1d.bias.shape)

conv1d.weight = nn.Parameter(linear.weight.unsqueeze(-1))
conv1d.bias = nn.Parameter(linear.bias)


Y_conv1d = conv1d(X.unsqueeze(-1)).squeeze(-1)

print(torch.max(torch.max(Y_conv1d - Y_linear)))
```

```python

import torch
import torch.nn as nn
num_samples = 10
input_dim = 5
output_dim = 3
seq_length = 20
X = torch.randn(size = (num_samples, seq_length, input_dim))
print(X.shape)
linear = nn.Linear(in_features=input_dim, out_features=output_dim)
print(linear.weight.shape)
print(linear.bias.shape)

Y_linear = linear(X)

conv1d = nn.Conv1d(in_channels=input_dim, out_channels=output_dim, kernel_size=1, stride=1, padding=0, bias=True)
print(conv1d.weight.shape)
print(conv1d.bias.shape)

conv1d.weight = nn.Parameter(linear.weight.unsqueeze(-1))
conv1d.bias = nn.Parameter(linear.bias)

Y_conv1d = conv1d(X.permute(0, 2, 1)).permute(0, 2, 1)

print(torch.max(torch.max(Y_conv1d - Y_linear)))
```

```python
import torch
import torch.nn as nn
num_samples = 10
input_dim = 5
output_dim = 3
seq_length = 20
time_window = 4
X = torch.randn(size = (num_samples, seq_length, input_dim))
print(X.shape)

conv1d = nn.Conv1d(in_channels=input_dim, out_channels=output_dim, kernel_size=time_window, stride=1, padding=time_window - 1, bias=True)
print(conv1d.weight.shape)
print(conv1d.bias.shape)

Y_conv1d = conv1d(X.permute(0, 2, 1)).permute(0, 2, 1)
print(Y_conv1d.shape)
```
