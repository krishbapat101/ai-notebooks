```python
import numpy as np
```

```python
class Affine:
    def __init__(self, weight, bias):
        self.weight = weight
        self.bias = bias

    def forward(self, x):
        return x @ self.weight.T + self.bias

    def __call__(self, x):
        return self.forward(x)
```

```python
class Identity:
  def __init__(self):
    pass

  def forward(self, x):
    return x

  def __call__(self, x):
    return self.forward(x)
```

```python
class Threshold:
    def __init__(self):
        pass

    def forward(self, x):
        return (x >= 0).astype(int)

    def __call__(self, x):
        return self.forward(x)
```

```python
class Sigmoid:
  def __init__(self):
    pass

  def forward(self, x):
    return 1/(1 + np.exp(-x))

  def __call__(self, x):
    return self.forward(x)
```

```python
class Tanh:
  def __init__(self):
    pass

  def forward(self, x):
    return (np.exp(x) - np.exp(-x))/(np.exp(x) + np.exp(-x))

  def __call__(self, x):
    return self.forward(x)
```

```python
class ReLU:
  def __init__(self):
    pass

  def forward(self, x):
    return np.maximum(0, x)

  def __call__(self, x):
    return self.forward(x)
```

```python
class GELU:
   def __init__(self):
     pass

   def forward(self, x):
     return x * 0.5 * (1 + np.erf(x / np.sqrt(2))) # erf ist the ge function

   def __call__(self, x):
     return self.forward(x)
```

```python
class MLP_AND:
    def __init__(self):
        self.affine1 = Affine(np.array([[1, 1]]), np.array([-2])) # (2, ) <- is vector vs (2, 1) <- is matrix for [1, 1] and [[1, 1]]
        self.activation1 = Threshold()
        self.layer1 = [self.affine1, self.activation1]
    def forward(self, x):

        return self.activation1(self.affine1(x)).astype(bool)

    def __call__(self, x):
        return self.forward(x)
```

```python
model_AND = MLP_AND()

n_samples = 10
np.random.seed(42)
x = np.random.randn(n_samples, 2) > 0
np.random.seed()
y = model_AND(x)
print('x', x, '\n')
print('shape:', x.shape, '\n')


print('x', y, '\n')
print('shape:', y.shape, '\n')
```

```python
class MLP_Triangle:
    def __init__(self):
        self.affine1 = Affine(np.array([[-1, 0],[0, -1],[1/np.sqrt(2), 1/np.sqrt(2)]]), np.array([0,0, -1/np.sqrt(2)]))
        self.activation1 = Threshold()
        self.layer1 = [self.affine1, self.activation1]

        self.affine2 = Affine(np.array([[-1, -1, -1]]), np.array([0]))
        self.activation2 = Threshold()
        self.layer2 = [self.affine2, self.activation2]
        self.layer = [self.layer1, self.layer2]
    def forward(self, x):
        for layer in self.layer:
            for module in layer:
                x = module(x)

        return x.astype(bool)
    def __call__(self, x):
        return self.forward(x)
```

```python
model_triangle = MLP_Triangle()


n_samples = 10
np.random.seed(42)
x = np.random.rand(n_samples, 2)
np.random.seed()
print('x', x, '\n')
print('shape:', x.shape, '\n')

y = model_triangle(x)
print('x', y, '\n')
print('shape:', y.shape, '\n')
```
