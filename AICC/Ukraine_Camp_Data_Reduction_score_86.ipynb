```python
import pandas as pd
train_data = pd.read_csv("train_data.csv")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier

model = Pipeline([
("imputer", SimpleImputer(strategy="median")),
("clf", DecisionTreeClassifier(random_state=42, class_weight="balanced", max_depth=None, min_samples_leaf=1)),
])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

X_train, X_test, y_train, y_test = train_test_split(train_data.drop(columns="class"), train_data["class"], test_size= 0.2, random_state=42)

```

```python
%%script false --no-raise-error
cols = [185, 141]
score = 0
for j in range(0):
    bestadd = -1
    score = 0
    for i in range((X_train.shape[1])):
        if i not in cols:
            cols.append(i)
            model.fit(X_train.iloc[:, cols], y_train)
            if score < f1_score(y_test, model.predict(X_test.iloc[:, cols]), average="macro"):
                score = f1_score(y_test, model.predict(X_test.iloc[:, cols]), average="macro")
                bestadd = i
            cols.pop()
    cols.append(bestadd)
    model.fit(X_train.iloc[:, cols], y_train)
    print(f1_score(y_test, model.predict(X_test.iloc[:, cols]), average="macro"))
    score = f1_score(y_test, model.predict(X_test.iloc[:, cols]), average="macro")
```

```python
import tqdm
```

```python
score = 0
cols = []
for j in tqdm.tqdm(range((X_train.shape[1]))):
    for i in range((X_train.shape[1])):

        testc = [i, j]
        model.fit(X_train.iloc[:, testc], y_train)
        if score < f1_score(y_test, model.predict(X_test.iloc[:, testc]), average="macro"):
            cols = testc
            score = f1_score(y_test, model.predict(X_test.iloc[:, cols]), average="macro")
```

```python
print(score)
print(cols)
X_train = X_train.iloc[:, cols].dropna()
X_test = X_test.iloc[:, cols]
```

```python
rows = [325]
cand = X_train.join(y_train).drop_duplicates().index
score = 0
while(score < 0.92):
    bestadd = -1
    score = 0
    for i in X_train.index:
        if i in cand:
            if i not in rows:
                rows.append(i)
                model.fit(X_train.loc[rows], y_train.loc[rows])
                if score < f1_score(y_test, model.predict(X_test), average="macro"):
                    score = f1_score(y_test, model.predict(X_test), average="macro")
                    bestadd = i
                rows.pop()
    rows.append(bestadd)
    model.fit(X_train.loc[rows], y_train.loc[rows])
    print(score)
```

```python
print(rows)
```

```python

sub = pd.DataFrame({"id": [0], "rows": [str(rows)], "columns": [str(cols)]})
sub.to_csv("submission.csv", index=False)
print(open("submission.csv").read())
```
