```python
import pandas as pd
import torch

df = pd.read_csv("data.csv")
y = df["y"].to_numpy(dtype="float32")
X = df.drop(columns="y").to_numpy(dtype="float32")

model = torch.jit.load("model.pt")
model.train(True)

with torch.no_grad():
    prediction = model(torch.from_numpy(X)).squeeze()
```

```python
model.train(True)
lr = 1e-7
z = 0
x.requires_grad = True
for i in tqdm.tqdm(range(1000)):
    x.requires_grad = True
    ((torch.norm((model(x) - y))) ** 2).backward()
    z -= x.grad[:,3] * lr
    x.grad.zero_()
    x.requires_grad = False
    x[:,3] += z


```
