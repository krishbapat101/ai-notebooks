```python
import numpy as np
import tqdm
import sklearn.preprocessing as lp
```

```python
matrix = df_train.values
```

```python
y_train0 = matrix[:, -1]
y_train0 = np.astype(y_train0, int)
matrix = np.delete(matrix, [-1], axis=1)
n_train = len(matrix)
print(matrix.shape)
```

```python
y_train = np.eye(3)[y_train0]
```

```python
matrix = np.vstack([matrix]+ [df_test.values])
```

```python
matrix = np.delete(matrix, [0], axis=1)
cols = matrix.T
print(cols[0][0])
```

```python
print(type(cols[0][0]))
print(cols[0][0][:4])
add = np.array([])
add = np.append(add, ([s[:4] for s in cols[0]]))

add = np.array(
    [[s[:4], s[5:7], s[8:10], s[11:13], s[14:16], s[17:19]] for s in cols[0]],
    dtype=int
)

print(add.shape, matrix.shape)   # (n, 6)
print(add[:3])

add = add.astype(float)
cols = np.vstack((cols, add.T))
cols = np.delete(cols, [0], axis = 0)
cols = cols.astype(float)
```

```python
data = cols.T.copy()
heavy = [8, 9]
data[:, heavy] = np.sign(data[:, heavy]) * np.log1p(np.abs(data[:, heavy]))
X_train = data[:n_train]
X_test  = data[n_train:]
col_mean = np.nanmean(X_train, axis=0)
X_train = np.where(np.isnan(X_train), col_mean, X_train)
X_test  = np.where(np.isnan(X_test),  col_mean, X_test)

mean = X_train.mean(axis=0)
std  = X_train.std(axis=0)
std[std == 0] = 1
X_train = (X_train - mean) / std
X_test  = (X_test  - mean) / std
X_train[:, -1] = 1
X_test[:, -1]  = 1
```

```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x))
```

```python
import matplotlib.pyplot as plt
a = 1e-3
history = []
weights1 = np.random.randn(len(X_train[0]), 12) * 0.1
weights2 = np.random.randn(12, 3) * 0.1
bias2 = np.zeros(3)
weights2 = np.astype(weights2, float)
weights1 = np.astype(weights1, float)
for epoch in tqdm.tqdm(range(300)):
    for i in range(len(X_train)):
        pred1 =  X_train[i] @ weights1
        z = sigmoid(pred1)
        pred2 =  z @ weights2 + bias2
        diff = -y_train[i] + pred2
        roll = weights2 @ diff
        roll = (1 - z) * z * roll
        weights2 -= a * np.outer(z, diff)
        bias2 -= a * diff
        weights1 -= a * np.outer(X_train[i], roll)
    history.append(np.mean((y_train - sigmoid(X_train @ weights1) @ weights2 - bias2) ** 2))
pd.Series(history).plot()
plt.show()
train_acc = (np.argmax((sigmoid(X_train @ weights1) @ weights2 + bias2), axis=1) == y_train0).mean()
print(train_acc)
```

```python
y_pred = np.argmax((sigmoid(X_test @ weights1) @ weights2 + bias2), axis=1)
```
