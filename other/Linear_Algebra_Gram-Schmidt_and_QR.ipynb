```python
import numpy as np
```

```python
x = np.eye(4) - np.diag([1, 1, 1], -1)
x
r2 = np.zeros((4,4))
print(x[:,0])
```

```python
a = x[:, 0].copy()
r2[:, 0] = [1, 0, 0, 0].copy()
```

```python
b = x[:, 1].copy()
c = x[:, 2].copy()
d = x[:, 3].copy()
```

```python
r2[:, 1] = [(a.T @ b)/(a.T @ a), 1, 0, 0].copy()
b -= (a.T @ b)/(a.T @ a) * a

b
```

```python
r2[:, 2] = [r2[1][0], (b.T @ c)/(b.T @ b), 1, 0].copy()
c -= (a.T @ c)/(a.T @ a) * a + (b.T @ c)/(b.T @ b) * b

c
```

```python
r2[:, 3] = [r2[1][0], r2[2,1], (c.T @ d)/(c.T @ c), 1].copy()
d -= (a.T @ d)/(a.T @ a) * a + (b.T @ d)/(b.T @ b) * b + (c.T @ d)/(c.T @ c) * c
d
```

```python
qo = np.array([a,b,c,d]).copy()
r1 = np.zeros((4,4))
for i in range(4):
    r1[i][i] = np.linalg.norm(qo[i]).copy()
a /= np.linalg.norm(a)
b /= np.linalg.norm(b)
c /= np.linalg.norm(c)
d /= np.linalg.norm(d)
```

```python
q = np.array([a, b, c, d]).copy()
q = q.T
r = (r1 @ r2)
print(((q @ r1) @ r2))
print((q @ r))
print(q)
print(r)

```
