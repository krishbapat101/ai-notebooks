```python
import numpy as np
import sklearn as skl
import sklearn.preprocessing as lp
```

```python
x = np.random.default_rng().integers(low=100, high=501, size=10)
x_2d = x.reshape(-1, 1)
```

```python
def minmaxnorm(x):
    x = x - x.min()
    x = x/ x.max()
    return x
```

```python
print(x)
print(minmaxnorm(x).flatten() -lp.MinMaxScaler().fit_transform(x_2d).flatten())
```

```python
def zscorenorm(x):
    x = x - np.mean(x)
    e = 0.000001
    x = x/(np.std(x) + e)
    return x
```

```python
print(x)
print(zscorenorm(x).flatten() - lp.StandardScaler().fit_transform(x_2d).flatten())
```

```python
price = ['high cost', 'mid cost', 'low cost']
toy = ['lego', 'board game', 'video game', 'trading card']

ns = 20

np.random.seed(42)
x = np.stack([np.random.choice(price, ns), np.random.choice(toy, ns)], axis=1)
print(x)
```

```python
class encoder:
    def __init__(self):
        self.classes = None
    def fit(self, X):
        self.classes = np.unique(X)
    def transform(self, X):
        X_encoded = np.zeros_like(X)
        for i, label in enumerate(self.classes):
            X_encoded[X == label] = i
        return X_encoded
labelero = lp.LabelEncoder()
encodermodelx =  np.empty_like(x)

labeler = encoder()
xt = x.T
encodermodelo = np.zeros((2, 20))
for i in range (x.shape[1]):
    labeler.fit(x[:, i])
    encodermodelx[:, i] = (labeler.transform(x[:, i]))
encodermodelx = encodermodelx.astype(float)
labelero.fit(xt[0])
encodermodelo[0] = labelero.transform(xt[0])
labelero.fit(xt[1])
encodermodelo[1] = labelero.transform(xt[1])
print(np.all(encodermodelx == encodermodelo.T))
```

```python
class onehot:
    def __init__(self):
        self.c = None
    def fit(self, X):
        self.c = []
        for i in range(X.shape[1]):
            self.c.append(np.unique(X[:, i]))
    def transform(self, X):
        numcol = 0
        for col in range(X.shape[1]):
            numcol += len(self.c[col])
        xonehot = np.zeros(shape=(X.shape[0], numcol))
        col_index = 0
        for col in range(X.shape[1]):
            for label in self.c[col]:
                xonehot[X[:, col] == label, col_index] = 1
                col_index += 1
        return xonehot

onehotm = onehot()
onehotm.fit(x)
print(onehotm.c)

ohm = onehotm.transform(x)
print(ohm)
```

```python
one_hot = lp.OneHotEncoder(sparse_output=False)
one_hot.fit(x)
print(one_hot.categories_)

one_hot_sklearn = one_hot.transform(x)
print(one_hot_sklearn)

# --- Compare our model and the one from Sklearn ---
print(np.all(ohm == one_hot_sklearn))
# Expected output: True
```
