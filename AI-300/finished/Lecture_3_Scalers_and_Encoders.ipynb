```python
class MinMaxScaler():
  def __init__(self) -> None:
    self._min = None
    self._max = None
  def fit_transform(self, X_trian: np.ndarray) -> np.ndarray:
    self._min = np.min(X_train, axis=0)
    self._max = np.max(X_train, axis=0)

    transformed_X = (X_train - self._min)/(self._max - self._min)
    return transformed_X
  def transform(self, X_test: np.ndarray) -> np.ndarray:
    transformed_X = (X_test - self._min)  / (self._max - self._min)
    return transformed_X
```

```python
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(scaler._min)
print(scaler._max)
```

```python
class StandarScaler():
  def __init__(self) -> None:
    self._mean = None
    self._std = None
  def fit_transform(self, X_trian: np.ndarray) -> np.ndarray:
    self._mean = np.mean(X_train, axis=0)
    self._std = np.std(X_train, axis=0)


    X_train_transformed = (X_train - self._mean) / (self._std)
    return X_train_transformed
  def transform(self, X_test: np.ndarray) -> np.ndarray:

    X_test_transformed = (X_test - self._mean) / (self._std)
    return X_test_transformed
```

```python
z_scaler = StandarScaler()

X_train_z_scaler = z_scaler.fit_transform(X_train)
X_test_z_scaler = z_scaler.transform(X_test)

print(z_scaler._mean)
print(z_scaler._std)
```

```python
color_labels = ['red', 'green', 'blue']
animal_labels = ['dog', 'cat', 'monkey', 'horse']

num_samples = 20

np.random.seed(42)
X = np.stack([np.random.choice(color_labels, size=num_samples),
              np.random.choice(animal_labels, size=num_samples)], axis = 1)
print(X)
```

```python
class LabelEncoder():
  def __init__(self) -> None:
    self._classes = None

  def fit(self, X: np.ndarray):
    self._classes = np.unique(X)

  def transform(self, X):
    X_encoded = np.zeros_like(X, dtype = int)
    for i, label in enumerate(self._classes):
      X_encoded[X == label] = i
    return X_encoded
```

```python
X_label_encoded_model = np.empty_like(X, dtype = int)

label_encoder = LabelEncoder()

for i in range(X.shape[1]):
  label_encoder.fit(X[:, i])
  print(f"All labels of fearure. {i} are: {label_encoder._classes}")
  X_label_encoded_model[:, i] = label_encoder.transform(X[:, i])

print(X_label_encoded_model)
```

```python
class OneHotEncoder():
  def __init__(self) -> None:
    self._categories = None

  def fit(self, X):
    self._categories = []  # list of labels of all columns
    for i in range(X.shape[1]):
      self._categories.append(np.unique(X[:, i]))
  def tansform(self, X):
    num_col_one_hot_encoded = 0
    for col in range(X.shape[1]):
      num_col_one_hot_encoded += len(self._categories[col])
    X_one_hot_encoded_model = np.zeros(shape=(X.shape[0], num_col_one_hot_encoded))

    col_index = 0
    for col in range(X.shape[1]):
      for label in self._categories[col]:
        X_one_hot_encoded_model[X[:,col] == label,col_index] = 1
        col_index += 1
    return X_one_hot_encoded_model
```

```python
one_hot_encoding_model = OneHotEncoder()
one_hot_encoding_model.fit(X)
print(one_hot_encoding_model._categories)

X_one_hot_encoding_model = one_hot_encoding_model.tansform(X)
print(X_one_hot_encoding_model)
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder(sparse_output = False)
one_hot_encoder.fit(X)
print(one_hot_encoder.categories_)

X_one_hot_encoding_sklearn = one_hot_encoder.transform(X)
print(X_one_hot_encoding_sklearn)
```
