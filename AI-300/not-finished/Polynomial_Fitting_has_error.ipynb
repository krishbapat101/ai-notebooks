```python
from typing import Literal
import numpy as np
from matplotlib import pyplot as plt
from sklearn. linear_model import LinearRegression
```

```python
def h_ground_truth(X):
  return 3 + X + np.power(X, 2)
```

```python
def model(model_degree: int, X_train: np.ndarray, y_train: np.ndarray) ->tuple[float, np.ndarray]:
  """fits a model of the given degree on the given training data."""
  phi_x_train = np.asarray([np.power(X_train, i) for i in range(1, model_degree + 1)])
  lin_reg = LinearRegression()
  lin_reg.fit(phi_x_train, y_train)

  return lin_reg.intercept_, lin_reg.coef_
```

```python
def prediction(intercept, coef, X_test, model_degree) -> np.ndarray:
  """Makes predictions with the given model on the given data."""
  phi_x_test = np.asarray([np.power(X_test, i) for i in range(1, model_degree + 1)]).transpose
  y_preds = intercept + phi_x_test @ coef
  return y_preds
```

```python


num_training_samples = 50

num_test_samples = 30
X_LB, X_UB = -3, 3
sigma = 2
```

```python
np.random.seed(42)

X_train_base = np.random.uniform(low=X_LB, high=X_UB, size=(num_training_samples,))
y_train_base = h_ground_truth(X_train_base) + np.random.normal(loc=0, scale=sigma, size = (num_training_samples, ))

X_test = np.random.uniform(0.7 * X_LB, 0.7 * X_UB, size=(num_test_samples, ))
y_test = h_ground_truth(X_test) + np.random.normal(loc=0, scale=sigma, size=(num_test_samples, ))
```

```python
model_degree_LB = 6
model_degree_UB = 8
from typing import Literal
def ensemple(num_models: int, data_generation_mode: Literal["bootstrap", "known"] = "bootstrap") -> np.ndarray:
  y_preds_all_models = np.zeros((num_models, num_test_samples))

  for n in range(num_models):
    if data_generation_mode == "known":
      X_train = np.random.uniform(low = X_LB, high = X_UB, size = (num_training_samples,))
    elif data_generation_mode == "bootstrap":
      indices = np.random.choice(np.arange(num_training_samples), size=(num_training_samples,), replace = True)
      X_train = X_train_base[indices]
      y_train = y_train_base[indices]
  model_degree = np.random.randint(model_degree_LB, model_degree_UB)
  intercept, coef = model(model_degree, X_train, y_train)
  curr_preds = prediction(intercept, coef, X_test, model_degree)
  y_preds_all_models[n] = curr_preds
  return y_preds_all_models
```
